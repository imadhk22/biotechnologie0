### Load `lab_data_test.csv`
Now, we will search for the file in your Drive and read it using pandas. If you know the exact folder path, you can update the path string directly.

```python
import os
import pandas as pd

# Search for the file in Google Drive
file_name = 'lab_data_test.csv'
file_path = None

for root, dirs, files in os.walk('/content/drive/MyDrive'):
    if file_name in files:
        file_path = os.path.join(root, file_name)
        print(f"Found file at: {file_path}")
        break

if file_path:
    df = pd.read_csv(file_path)
    display(df.head())
else:
    print(f"Could not find '{file_name}' in your Google Drive. Please make sure the spelling is correct and that it is uploaded to your Drive.")
```
